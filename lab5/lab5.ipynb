**Zadanie 1** Napisz program, który znajdzie w pliku `imiona.txt` wszystkie takie imiona, które zawierają literę `ą`. Skorzystaj z funkcji `re.search`

```python
import re

with open('imiona.txt') as plik:
    for linia in plik:
        if re.search('ą', linia):
            print(linia.strip())
```

**Zadanie 14** Napisz program, który znajdzie w pliku `imiona.txt` wszystkie takie imiona, które są palindromami długości 4 liter. Palindrom to taki wyraz, który czytany wspak jest taki sam, jak czytany w przód. Użyj odwołań wstecznych i kotwic. Koniecznie użyj też metody `.lower` lub `.upper`, aby móc sprawdzać, czy pierwsza litera jest równa ostatniej

```python
with open('imiona.txt') as plik:
    for linia in plik:
        if re.search(r'^(.)(.)\2\1$', linia.lower()):
            print(linia.strip())
```

**Zadanie 15** Napisz funkcję `czy_numer_telefonu_jest_poprawny(numer: str) -> bool`. Ta funkcja ma zwracać `True` tylko wtedy, gdy argument `numer` może być numerem telefonu komórkowego, czyli gdy ma format `+48 XXX-XXX-XXX`. Użyj znaku ucieczki, aby regexp poprawnie rozpoznawał znak `+`

```python
def czy_numer_telefonu_jest_poprawny(numer: str) -> bool:
    if re.search(r'^\+48 [0-9]{3}-[0-9]{3}-[0-9]{3}$', numer):
        return True
    return False
```

```python
assert(czy_numer_telefonu_jest_poprawny('+48 123-456-789'))
assert(not czy_numer_telefonu_jest_poprawny('+48 123-456-78x'))
assert(not czy_numer_telefonu_jest_poprawny('+48 123-456 789'))
assert(not czy_numer_telefonu_jest_poprawny('+48  123-456-789'))
assert(not czy_numer_telefonu_jest_poprawny('48 123-456-789'))
assert(not czy_numer_telefonu_jest_poprawny(' +48 123-456-789'))
assert(not czy_numer_telefonu_jest_poprawny('+48 123-456-78'))
assert(not czy_numer_telefonu_jest_poprawny('+48 123-456-7890'))
```

Każdy gen DNA mitochondrialnego kręgowców składa się z:

* Dowolnej liczby kodonów, które kodują kolejne aminokwasy. Każdy z tych kodonów to trzy nukleotydy

Proszę pamiętać, że DNA mitochondrialne kręgowców różni się od RNA:
* w mDNA jest tymina, a w RNA jest uracyl
* w mDNA są inne warianty kodonów START i STOP

Proszę przypisać do zmiennej `gen_re` taki regexp, który opisuje jeden gen DNA mitochondrialnego kręgowców. Proszę skorzystać z niezachłannego kwantyfikatora

```python
mdna = 'TGGGCATTGATGCGTACTGGAAATGCTGATTAAGCCAAATGTTTGATGAGACCTGGGAACGCCGACTAA'
gen_re = ''
```

```python
geny = re.findall(gen_re, mdna)
print(geny)
assert(len(geny) == 2)
```
