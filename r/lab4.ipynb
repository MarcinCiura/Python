**Zadanie**. Napisz funkcję `jest_zaczepką(i1: str, i2: str) -> bool`. Ta funkcja ma zwracać
`True`, jeśli `i2` jest zaczepką imienia `i1`.
Jeśli `i2` nie jest zaczepką imienia `i1`, ta funkcja ma zwracać `False`. Całe `i1` nie jest swoją zaczepką.

Wskazówka: Użyj pętli, w której zmienna `i` przyjmuje wartości od 2 do `len(i2)` włącznie. Jeśli `i1` kończy się na wycinek `i2[:i]`, niech ta funkcja zwróci `True`. Jeśli ta pętla zakończy się bez zwracania `True`, niech ta funkcja zwróci `False`.

````python
def jest_zaczepką(i1: str, i2: str) -> bool:
    """Zwraca True, jeśli i2 jest dokończeniem i1."""
    for i in range(2, len(i2) + 1):
        if i1[1:].endswith(i2[:i]):
            return True
    return False
````

````python
assert jest_zaczepką('ANNA', 'NATALIA'), f'{jest_zaczepką('ANNA', 'NATALIA')}'
assert jest_zaczepką('ALAN', 'LANCELOT'), f'{jest_zaczepką('ALAN', 'LANCELOT')}'
assert not jest_zaczepką('PIOTR', 'PIOTR'), f'{jest_zaczepką('PIOTR', 'PIOTR')}'
assert not jest_zaczepką('ANNA', 'PIOTR'), f'{jest_zaczepką('ANNA', 'PIOTR')}'
assert not jest_zaczepką('PIOTR', 'PIOTR'), f'{jest_zaczepką('PIOTR', 'PIOTR')}'
assert not jest_zaczepką('PIOTR', 'PIOTREK'), f'{jest_zaczepką('PIOTR', 'PIOTREK')}'
assert jest_zaczepką('JOANNA', 'ANNA'), f'{jest_zaczepką('JOANNA', 'ANNA')}'
````

którego kluczem będzie imię, a wartością lista zaczepek tego imienia.

Wskazówka:
Utwórz pusty słownik `zaczepki`.
Użyj pętli, w której zmienna `i1` przebiega po elementach listy `imiona`. Wewnątrz tej pętli przypisz do `zaczepki[i1]` listę. W tej liście znajdą się takie imiona `i2` z listy `imiona`, że `jest_zaczepką(i1, i2)`. Na końcu tej funkcji zwróć `zaczepki`.

````python
def twórz_zaczepki(imiona: list) -> dict:
    """Zwraca słownik zaczepek"""
    zaczepki = {}
    for i1 in imiona:
        zaczepki[i1] = [i2 for i2 in imiona if jest_zaczepką(i1, i2)]
    return zaczepki
````

**Zadanie**. Napisz funkcję `wybierz_zaczepkę(imię: str) -> str | None`. Ta funkcja zwraca losowe dokończenie imienia pobrane ze słownika `zaczepki` lub `None`, gdy `imię` nie ma dokończenia.

Wskazówka: znajdź w zaczepkach tę wartość, która odpowiada imieniu. Ta wartość to lista. Gdy ta lista nie jest pusta, wybierz z tej listy losowy element. W przeciwnym przypadku domyśl się, co robić :-)

````python
import random

def wybierz_zaczepkę(imię: str) -> str | None:
    """Zwraca zaczepkę imienia lub `None`"""
    z = zaczepki[imię]
    if z:
        return random.choice(z)
    else:
        return None
````

````python
assert(wybierz_zaczepkę('ADAŚ') == 'AŚKA'), f'{wybierz_zaczepkę('ADAŚ')}'
assert(wybierz_zaczepkę('WILHELM') in {'HELMUT', 'WILHELMINA'}), f'{wybierz_zaczepkę('WILHELM')}'
assert(wybierz_zaczepkę('NATALIA') is None), f'{wybierz_zaczepkę('NATALIA')}'
````
